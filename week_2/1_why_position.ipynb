```python
import torch

torch.manual_seed(0)
X = torch.randn(3, 4)                       # 3 token: dog, bites, man; shape (3, 4)
```

```python
def self_attn(X):                           # Q=K=V=X, basit hali
    scores = X @ X.T / (X.shape[-1] ** 0.5) # (3,3)
    return torch.softmax(scores, -1) @ X    # (3,4)

out1 = self_attn(X)
perm = [2, 1, 0]                            # "man bites dog"
out2 = self_attn(X[perm])

# out2, out1'in aynı permütasyonla yeniden sıralanmışı mı?
print(torch.allclose(out2, out1[perm], atol=1e-6))
```
