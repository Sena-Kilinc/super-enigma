```python
import torch
import torch.nn.functional as F

n = 4
# Bidirectional: hepsi 0 (engelleme yok)
mask_bi = torch.zeros(n, n)
# Causal: üst üçgen (köşegen hariç) -inf
mask_causal = torch.triu(torch.full((n, n), float("-inf")), diagonal=1)
print(mask_causal)

# Dosyadaki örnek: skorlar [1,2,1,2], maske [0,0,-inf,-inf]
s = torch.tensor([1., 2., 1., 2.])
m = torch.tensor([0., 0., float("-inf"), float("-inf")])
print("softmax:", torch.softmax(s + m, -1).round(decimals=2))

# Tüm matris: causal maske ile attention ağırlıkları
torch.manual_seed(0)
scores = torch.randn(n, n)
w_bi = F.softmax(scores + mask_bi, -1)
w_ca = F.softmax(scores + mask_causal, -1)
print("causal ağırlıklar:\n", w_ca.round(decimals=2))
print("satır toplamları:", w_ca.sum(-1))
```
