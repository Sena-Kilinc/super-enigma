```python
import torch
import torch.nn as nn
```

```python
# --1) Kelime Düzeyinde Token -> ID
vocab = {"the": 0, "dog": 1, "sat": 2, "on": 3, "mat": 4, ".": 5}
text_tokens = ["the", "dog", "sat", "on", "the", "mat", "."]
ids = torch.tensor([vocab[t] for t in text_tokens])  # Convert tokens to IDs shape: (7,)
print("Token IDs:", ids.tolist())
```

```python
# --2) Embedding /Gömleme Tablosu: V x d_model

V, d_model= len(vocab), 4  # Vocabulary size and embedding dimension
emb = nn.Embedding(V, d_model)  # Create an embedding layer ağırlık shape: (6, 4)
vectors = emb(ids)  # Convert IDs to embeddings shape: (7, 4)
print("embedding tablosu:", emb.weight.shape, "| çıktı:",vectors.shape)
```

```python
# --3) Alt-kelime: Greedy en uzun eşleşme
sub_vocab = {"un", "believ", "able", "ability", "ity"}
def subword_tokenize(word):
    out, i = [], 0
    while i < len(word):
        for j in range(len(word), i, -1):                # en uzun parçadan başla
            if word[i:j] in sub_vocab:
                out.append(word[i:j])
                i = j
                break
        else:
            out.append(word[i]); i += 1                  # bilinmeyen: tek harf
            i += 1
    return out
print("Subword tokenization of 'unbelievable':", subword_tokenize("unbelievability"))
```
