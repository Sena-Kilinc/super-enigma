```python
from collections import Counter
```

```python
corpus = ("the dog sat on the mat . "
          "the dog sat on the floor . "
          "the cat sat on the mat .")
tokens = corpus.split()  # Split the corpus into tokens
print(tokens)
```

```python
def ngram_probs(tokens, context):
    """context: önceki n-1 kelimenin listesi. Dönüş:  {sonraki kelime: olasılık}"""
    k = len(context) # açıklama: context listesinde kaç kelime olduğunu buluyoruz ve k değişkenine atıyoruz
    counts = Counter() # açıklama: Counter sınıfını kullanarak bir sayıcı oluşturuyoruz
    for i in range(len(tokens) - k): # açıklama: tokens listesinde k kelimeyi geçmeden tüm kelimeleri dolaşıyoruz
        if tokens[i:i+k] == list(context): # açıklama: tokens listesinde i'den i+k'ya kadar olan kelimeler context ile aynı mı diye kontrol ediyoruz
            counts[tokens[i+k]] += 1 # açıklama: eğer aynı ise counts sayıcısında tokens[i+k] kelimesinin sayısını bir artırıyoruz
    total = sum(counts.values()) # açıklama: counts sayıcısındaki tüm değerlerin toplamını buluyoruz ve total değişkenine atıyoruz
    if total == 0: # açıklama: eğer total 0 ise yani context ile eşleşen kelime yoksa boş bir sözlük döndürüyoruz
        return {}
    return {w:round(c/total, 4) for w, c in counts.items()} # açıklama: counts sayıcısındaki her kelime ve sayısını alıyoruz ve her kelimenin olasılığını hesaplayarak bir sözlük döndürüyoruz
print("Bigram P(.| the) :",ngram_probs(tokens, ["the"])) # açıklama: ngram_probs fonksiyonunu çağırıyoruz ve context olarak ["the"] veriyoruz ve bigram olasılıklarını yazdırıyoruz
print("Trigram P(.| on the) :",ngram_probs(tokens, ["on", "the"])) # açıklama: ngram_probs fonksiyonunu çağırıyoruz ve context olarak ["on", "the"] veriyoruz ve trigram olasılıklarını yazdırıyoruz
print("Sparsity P(.|quantum banana) :",ngram_probs(tokens, ["quantum", "banana"])) # açıklama: ngram_probs fonksiyonunu çağırıyoruz ve context olarak ["quantum", "banana"] veriyoruz ve sparsity olasılıklarını yazdırıyoruz
```
