```python
import torch
import torch.nn.functional as F
```

```python
# Tek örnek: 3 adaylı sözlük, doğru token indeksi 0 ("mat")
logits = torch.tensor([[3.0, 1.0, 0.0]])       # shape: (1, 3)  (batch=1, V=3)
target = torch.tensor([0])                      # shape: (1,)
```

```python
probs = torch.softmax(logits, dim=-1)
manual_loss = -torch.log(probs[0, target.item()])
torch_loss  = F.cross_entropy(logits, target)   # softmax + -log birlikte !!! F.cross_entropy içinde softmax vardır, ona logit verilir (olasılık değil).
print("manuel kayıp:", manual_loss.item())
print("torch kayıp :", torch_loss.item())
```
