📄 Örnek: “The dog sat on the mat”:
P = P(The)·P(dog|The)·P(sat|The,dog)·P(on|The,dog,sat)·P(the|…)·P(mat|…)

➕ Sayısal örnek (uydurma olasılıklarla): Altı çarpan: 0.2, 0.5, 0.4, 0.6, 0.7, 0.3.
0.2×0.5 = 0.1 → ×0.4 = 0.04 → ×0.6 = 0.024 → ×0.7 = 0.0168 → ×0.3 = 0.00504.

Uzun dizilerde bu çarpım çok küçülür (underflow). Bu yüzden pratikte log toplamı kullanılır: log P = Σ log pᵢ.

```python
import math
```

```python
ps = [0.2, 0.5, 0.4, 0.6, 0.7, 0.3]

prod = 1.0
for p in ps:
    prod *= p                                   # doğrudan çarpım

log_sum = sum(math.log(p) for p in ps)          # log toplamı

print("çarpım: ", round(prod, 5))
print("log toplamı: ", round(log_sum, 4))
print("exp(log toplamı): ", round(math.exp(log_sum), 5))
```

Yorum: İki yol aynı sonucu verir. Log yolu uzun dizilerde sayısal olarak güvenlidir. Eğitimdeki kayıp fonksiyonunun (−log P) neden log kullandığının nedeni de budur.
