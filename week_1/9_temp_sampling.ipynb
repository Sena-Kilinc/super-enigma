```python
import torch

torch.manual_seed(42)
tokens = ["mat", "floor", "roof"]
z = torch.tensor([3.0, 1.0, 0.0])                 # (3,)
```

```python
def with_temperature(z, T):
    return torch.softmax(z / T, dim=-1)           # (3,)

for T in [0.5, 1.0, 2.0]:
    print(f"T={T}:", with_temperature(z, T).round(decimals=4))

# Greedy
p = with_temperature(z, 0.5)
print("greedy:", tokens[p.argmax().item()])

# Sampling (T=2.0 ile 1000 çekiliş, sıklıkları say)
p2 = with_temperature(z, 2.0)
draws = torch.multinomial(p2, 1000, replacement=True)      # (1000,)
freq = torch.bincount(draws, minlength=3) / 1000
print("T=2.0 çekiliş sıklıkları:", dict(zip(tokens, freq.tolist())))
```
