Formül: pᵢ = exp(zᵢ) / Σⱼ exp(zⱼ)
Büyük logitlerde exp taşar (örn. e¹⁰⁰⁰ = ∞). Çözüm: her logitten maksimumu çıkarmak. Sonuç matematiksel olarak aynıdır, çünkü pay ve paydadaki e^(−max) çarpanı sadeleşir. Hafta 2'deki kodda x - np.max(x) satırı budur.

```python
import torch

z = torch.tensor([3.0, 1.0, 0.0])            # logitler, shape: (3,)
```

```python
# Elle adım adım
e = torch.exp(z)                              # (3,)
p_manual = e / e.sum()                        # (3,)

# Hazır fonksiyon
p_torch = torch.softmax(z, dim=-1)            # (3,)

print("exp   :", e)
print("toplam:", e.sum().item())
print("manuel:", p_manual)
print("torch :", p_torch)
print("p toplamı:", p_torch.sum().item())
```

```python
# Kararlılık testi
big = torch.tensor([1000.0, 999.0, 998.0])
naive = torch.exp(big) / torch.exp(big).sum()               # taşar
stable = torch.exp(big - big.max()) / torch.exp(big - big.max()).sum()
print("naive :", naive)
print("stable:", stable)
```
