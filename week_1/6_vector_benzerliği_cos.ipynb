```python
import torch
import torch.nn.functional as F
```

```python
cat  = torch.tensor([0.9, 0.8])
dog  = torch.tensor([0.8, 0.6])
king = torch.tensor([-0.2, 0.9])
```

```python
print("cos(cat, dog) =", round(F.cosine_similarity(cat, dog, dim=0).item(), 4))
print("cos(cat, king) =", round(F.cosine_similarity(cat, king, dim=0).item(), 4))
```
