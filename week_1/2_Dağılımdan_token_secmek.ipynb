```python
import torch
```

```python
torch.manual_seed(0) # Set the random seed for reproducibility
tokens= ["mat","floor","roof"]
prob = torch.tensor([0.84, 0.12, 0.04]) # Define the probabilities for each token shape (3,) toplam=1.00

# greedy: en yüksek olasılıklı olanı seç
print("Greedy sampling:", tokens[torch.argmax(prob).item()]) #açıklama: torch.argmax(prob) ile prob tensoründeki en yüksek olasılıklı tokenın indeksini buluyoruz ve tokens listesinde bu indeksi kullanarak tokenı seçiyoruz ve yazdırıyoruz

#sampling: olasılığa göre rastgele seç
idx = torch.multinomial(prob,num_samples=10,replacement=True) # açıklama: num_samples=10, replacement=True ile 10 örnek seçiyoruz ve tekrar seçime izin veriyoruz multinomial fonksiyonu ile olasılığa göre rastgele seçim yapıyoruz
print("Sampling:", [tokens[i] for i in idx]) # açıklama: idx ile seçilen tokenların indekslerini kullanarak tokens listesinde bu indeksleri kullanarak tokenları seçiyoruz ve liste olarak yazdırıyoruz
```
